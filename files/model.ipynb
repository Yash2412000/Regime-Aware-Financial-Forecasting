# Model architecture
Imports `model.py` and runs a forward-pass sanity check with dummy data.

```
import torch
from model import MoEForecaster
from dataset import SEQ_LEN, FEATURE_COLS

N_FEATURES = len(FEATURE_COLS)
B, T, F = 8, SEQ_LEN, N_FEATURES
x_dummy = torch.randn(B, T, F)

model = MoEForecaster()
y_hat, g_w, preds = model(x_dummy)

print(f'Experts: {model.expert_names}')
print(f'y_hat: {y_hat.shape}, gate weights: {g_w.shape}, expert preds: {preds.shape}')
print(f'Gate sums (should be ~1.0): {g_w.sum(dim=-1)}')
print(f'Parameter count: {sum(p.numel() for p in model.parameters()):,}')
```
